# GA Final Models Comparison

This notebook trains Random Forest and XGBoost models using the features selected by the Genetic Algorithm, and compares their performance against the baseline models without feature selection.

```python
import os
import sys
import json
import pandas as pd
from IPython.display import Image, display

project_root = os.path.abspath('../../')
if project_root not in sys.path:
    sys.path.append(project_root)
```

```python
# Run the training script
!python ../../src/train_ga_models.py
```

```python
# Load and display the results as a dataframe for the table
results_dir = os.path.join(project_root, 'results')
ga_models_path = os.path.join(results_dir, 'ga_models.json')
baseline_rf_path = os.path.join(results_dir, 'baseline_rf.json')

with open(ga_models_path, 'r') as f:
    ga_data = json.load(f)
    
try:
    with open(baseline_rf_path, 'r') as f:
        baseline_data = json.load(f)
except FileNotFoundError:
    baseline_data = {"RMSE": 0, "MAE": 0, "R2": 0, "training_time": 0, "n_features_used": 0}

comparison = {
    "Model": ["Baseline RF", "GA + RF", "GA + XGBoost"],
    "Features Used": [baseline_data.get('n_features_used', 9), ga_data['ga_rf']['n_features'], ga_data['ga_xgb']['n_features']],
    "RMSE": [baseline_data['RMSE'], ga_data['ga_rf']['rmse'], ga_data['ga_xgb']['rmse']],
    "MAE": [baseline_data['MAE'], ga_data['ga_rf']['mae'], ga_data['ga_xgb']['mae']],
    "R2": [baseline_data['R2'], ga_data['ga_rf']['r2'], ga_data['ga_xgb']['r2']],
    "Train Time (s)": [baseline_data.get('training_time', 0), ga_data['ga_rf']['training_time'], ga_data['ga_xgb']['training_time']]
}

df_comp = pd.DataFrame(comparison)
display(df_comp)
```

```python
# Display the comparison chart
plot_path = os.path.join(results_dir, 'ga_vs_baseline_comparison.png')
display(Image(filename=plot_path))
```
